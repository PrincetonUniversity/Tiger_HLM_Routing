```python
import pandas as pd
import numpy as np

df = pd.read_csv("/scratch/gpfs/GVILLARI/mb6477/AORC_1km_CONUS/Tiger_HLM_tutorial/inputs/parameters/routing_table_Raritan.csv")

# parents references the 'index' column, semicolon-separated, NaN for headwaters
def parse_parents(p):
    if pd.isna(p):
        return []
    return [int(x) for x in str(p).split(";")]

df["parent_list"] = df["parents"].apply(parse_parents)

# params: A_h;L_i;lambda_1;v_0
param_cols = df["params"].str.split(";", expand=True).astype(float)
df["A_h"] = param_cols[0]
df["L_i"] = param_cols[1]
df["lambda_1"] = param_cols[2]
df["v_0"] = param_cols[3]

print(f"Total nodes: {len(df)}")
print(f"Headwaters: {(df['level']==0).sum()}")
print(f"Max level: {df['level'].max()}")
```

```python
level_counts = df["level"].value_counts().sort_index()
single_node_levels = (level_counts == 1).sum()
print(f"Levels with exactly 1 node: {single_node_levels} / {len(level_counts)}")
print(f"Total barrier-synchronizations that do essentially nothing: ~{single_node_levels}")

import matplotlib.pyplot as plt
plt.figure(figsize=(10,4))
plt.plot(level_counts.index, level_counts.values)
plt.yscale("log")
plt.xlabel("Level")
plt.ylabel("Nodes at level (log scale)")
plt.title("Raritan network: nodes per level")
plt.axhline(1, color="red", linestyle="--", label="1 node")
plt.legend()
plt.show()
```

```python
# Paste your full level_analysis.txt per-level totals here (level -> (total_time_53chunks, nodes_per_chunk))
level_stats = {
    0: (6.2720, 9015), 1: (1.3169, 1800), 2: (1.3851, 998),
    3: (0.9081, 656), 4: (0.7398, 485), 5: (0.6061, 381),
    6: (0.4658, 304), 7: (0.4505, 251), 8: (0.3075, 209),
    11: (0.3049, 139),
    # add the rest of your levels here from level_analysis.txt
}

avg_node_cost_by_level = {
    lvl: (total / 53) / nodes for lvl, (total, nodes) in level_stats.items()
}
global_avg_cost = np.mean(list(avg_node_cost_by_level.values()))

def node_cost(level):
    return avg_node_cost_by_level.get(level, global_avg_cost)

df["est_cost"] = df["level"].apply(node_cost)
```

```python
from collections import deque

n = len(df)
parents_of = df["parent_list"].to_dict()  # index -> list of parent indices
children_of = {i: [] for i in range(n)}
for i, plist in parents_of.items():
    for p in plist:
        children_of[p].append(i)

in_degree = {i: len(parents_of[i]) for i in range(n)}
queue = deque([i for i in range(n) if in_degree[i] == 0])
topo_order = []
while queue:
    i = queue.popleft()
    topo_order.append(i)
    for c in children_of[i]:
        in_degree[c] -= 1
        if in_degree[c] == 0:
            queue.append(c)

assert len(topo_order) == n, f"Only {len(topo_order)}/{n} nodes reached — check for cycles"

cost = df["est_cost"].values
weighted_dist = np.zeros(n)
predecessor = np.full(n, -1, dtype=int)

for i in topo_order:
    plist = parents_of[i]
    if not plist:
        weighted_dist[i] = cost[i]
    else:
        best_p = max(plist, key=lambda p: weighted_dist[p])
        weighted_dist[i] = weighted_dist[best_p] + cost[i]
        predecessor[i] = best_p

critical_node = int(np.argmax(weighted_dist))
critical_path_time = weighted_dist[critical_node]

print(f"Critical path (estimated): {critical_path_time:.3f}s per chunk")
print(f"x53 chunks: {critical_path_time * 53:.2f}s")
print(f"Measured total: 71.754s")
print(f"Headroom above critical path: {71.754 - critical_path_time*53:.2f}s ({100*(1 - critical_path_time*53/71.754):.1f}%)")
```

```python
path = []
i = critical_node
while i != -1:
    path.append(i)
    i = predecessor[i]
path.reverse()

print(f"Critical path length: {len(path)} nodes")
print(f"Critical path node IDs (first 5, last 5):")
print(df.loc[path[:5], "node"].tolist(), "...", df.loc[path[-5:], "node"].tolist())
```
